```python
def load_ohlc_file(filepath):
    df = pd.read_csv(filepath)
    
    # Standardise column names
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
    )
    
    return df
```

```python
dataframes = []

for file in csv_files:
    df = load_ohlc_file(file)
    df["source_file"] = file.name
    dataframes.append(df)

data = pd.concat(dataframes, ignore_index=True)
```

```python
def calculate_supertrend(
    data,
    atr_period=10,
    multiplier=3.0
):

    df = data.copy()

    # --------------------------------------------------
    # 1. True Range
    # --------------------------------------------------

    previous_close = df["close"].shift(1)

    tr = pd.concat(
        [
            df["high"] - df["low"],
            (df["high"] - previous_close).abs(),
            (df["low"] - previous_close).abs()
        ],
        axis=1
    ).max(axis=1)

    # --------------------------------------------------
    # 2. Wilder ATR
    # --------------------------------------------------

    atr = tr.ewm(
        alpha=1 / atr_period,
        adjust=False,
        min_periods=atr_period
    ).mean()

    # --------------------------------------------------
    # 3. Basic bands
    # --------------------------------------------------

    hl2 = (
        df["high"] +
        df["low"]
    ) / 2

    basic_upper = (
        hl2 +
        multiplier * atr
    )

    basic_lower = (
        hl2 -
        multiplier * atr
    )

    # --------------------------------------------------
    # 4. Convert to NumPy arrays
    # --------------------------------------------------

    close = df["close"].to_numpy()
    basic_upper = basic_upper.to_numpy()
    basic_lower = basic_lower.to_numpy()
    atr = atr.to_numpy()

    n = len(df)

    final_upper = np.full(n, np.nan)
    final_lower = np.full(n, np.nan)
    supertrend = np.full(n, np.nan)
    trend = np.full(n, np.nan)

    # --------------------------------------------------
    # 5. First valid ATR observation
    # --------------------------------------------------

    valid_indices = np.where(~np.isnan(atr))[0]

    first_valid = valid_indices[0]

    # Initialize bands
    final_upper[first_valid] = basic_upper[first_valid]
    final_lower[first_valid] = basic_lower[first_valid]

    # Initial trend
    if close[first_valid] <= final_upper[first_valid]:

        trend[first_valid] = -1

        supertrend[first_valid] = (
            final_upper[first_valid]
        )

    else:

        trend[first_valid] = 1

        supertrend[first_valid] = (
            final_lower[first_valid]
        )

    # --------------------------------------------------
    # 6. Recursive Supertrend calculation
    # --------------------------------------------------

    for i in range(first_valid + 1, n):

        # Final upper band
        if (
            basic_upper[i] < final_upper[i - 1]
            or close[i - 1] > final_upper[i - 1]
        ):

            final_upper[i] = basic_upper[i]

        else:

            final_upper[i] = final_upper[i - 1]

        # Final lower band
        if (
            basic_lower[i] > final_lower[i - 1]
            or close[i - 1] < final_lower[i - 1]
        ):

            final_lower[i] = basic_lower[i]

        else:

            final_lower[i] = final_lower[i - 1]

        # Trend transition
        if trend[i - 1] == -1:

            if close[i] > final_upper[i]:

                trend[i] = 1

                supertrend[i] = final_lower[i]

            else:

                trend[i] = -1

                supertrend[i] = final_upper[i]

        else:

            if close[i] < final_lower[i]:

                trend[i] = -1

                supertrend[i] = final_upper[i]

            else:

                trend[i] = 1

                supertrend[i] = final_lower[i]

    # --------------------------------------------------
    # 7. Add results to DataFrame
    # --------------------------------------------------

    df["tr"] = tr
    df["atr"] = atr
    df["basic_upper_band"] = basic_upper
    df["basic_lower_band"] = basic_lower
    df["final_upper_band"] = final_upper
    df["final_lower_band"] = final_lower
    df["supertrend"] = supertrend
    df["trend"] = trend

    return df



window_data = calculate_supertrend(
    data.iloc[0:40000],
    atr_period=10,
    multiplier=3.0
)
```
